# LISTAS

```python
# Criando lista de cidades
cidades = ["Florianopolis", "Criciuma", "Blumenau", "Joinville", "Timbo"]

print(cidades[2])
```

### Adicionando elementos na lista

```python
# Adicionando elementos com o comando append
cidades.append("Jaragua do Sul")
cidades.append("Tubarao")
```

```python
print(cidades)
```

```python
# Adicionando elementos com o comando Insert

cidades.insert (1, "Balneario Camburiu")
print(cidades)
```

### Removendo elementos da lista

```python
# Usando o comando remove
cidades.remove ("Timbo")
print (cidades)
```

```python
# Usando o comando pop
removido = cidades.pop(3)
print (f"Cidade removida: {removido}")
print(cidades)
```

```python
# Usando o comando del
del cidades[3]
print(cidades)
```

```python
# Quantidade de elementos na lista
print(len(cidades))
```

### Fatiamento em listas

```python
print(cidades[1:5])
```

```python
print(cidades[:4])
```

### Substituição de Valores em listas

```python
print(cidades) #aparece todas que tenho
```
