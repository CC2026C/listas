```python
compras = ["arroz", "feijão", "leite", "café", "açúcar"]
```

```python
print("Lista inicial:")
print(compras)
```

```python
#Adicionar um produto - Macarrão
compras.append("macarrão")
print(compras)
```

```python
#Inserir o produto 'óleo' na segunda posição da lista
compras.insert(2, "óleo")
print(compras)
```

```python
#Remover o produto 'leite'
compras.remove ("leite")
print(compras)
```

```python
#Remover o último produto 'macarrão' com pop
produto_removido = compras.pop(5)
print(f"Produto removido: {produto_removido}")
```

```python
#Quantidade de produtos
print(f"Quantidade de produtos: {len(compras)}")
```

```python
#Indique os 3 primeiros produtos com o Fatiamemto
print("Primeiros três produtos:")
print(compras[0:3])

```

```python
#Substitua 'arroz' por 'achocolatado'
#Alterando um valor
compras[3] = "achocolatado"
print(compras)
```

```python
#Utilize o 'for' para exibir todos os produtos da lista
print("lista final:")

for produto in compras: 
    print (produto)
```
